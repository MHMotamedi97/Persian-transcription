```python
from pydub import AudioSegment
import os

def split_audio(input_file, chunk_length_ms=60000):
    audio = AudioSegment.from_file(input_file)
    chunks = [audio[i:i + chunk_length_ms] for i in range(0, len(audio), chunk_length_ms)]

    base_name = os.path.splitext(os.path.basename(input_file))[0]
    output_dir = f"{base_name}_chunks"
    os.makedirs(output_dir, exist_ok=True)

    for i, chunk in enumerate(chunks):
        chunk_name = os.path.join(output_dir, f"{base_name}_part{i + 1}.wav")
        chunk.export(chunk_name, format="wav")
        print(f"Saved: {chunk_name}")

split_audio("./Capture 4-16-25(1).WAV")
```

```python
import os

def count_files(directory):
  """Counts all files within a specified directory.

  Args:
      directory: The path to the directory.

  Returns:
      The total number of files in the directory.
  """
  total_files = 0
  for _, _, files in os.walk(directory):
    total_files += len(files)
  return total_files

directory_path = "/content/Capture 4-16-25(1)_chunks"
file_count = count_files(directory_path)
print(f"Total number of files in '{directory_path}': {file_count}")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage
import base64
import time

llm = ChatGoogleGenerativeAI(model="gemini-2.0-flash")
text = ""
for i in range(1, file_count+1):
  audio_file_path = f"/content/Capture 4-16-25(1)_chunks/Capture 4-16-25(1)_part{i}.wav"
  audio_mime_type = "audio/wav"

  with open(audio_file_path, "rb") as audio_file:
      encoded_audio = base64.b64encode(audio_file.read()).decode('utf-8')

  message = HumanMessage(
      content=[
          {"type": "text", "text": "Transcribe this audio."},
          {"type": "media", "data": encoded_audio, "mime_type": audio_mime_type}
      ]
  )

  response = llm.invoke([message])
  print(response.content)
  text = text + response.content
  time.sleep(10)
```

```python
with open("output.txt", "w") as file:
  file.write(text)
```
